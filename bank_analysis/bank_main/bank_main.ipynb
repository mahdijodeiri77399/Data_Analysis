تست‌های آماری
عددی به کتگوری

    T-test: وقتی متغیر عددی تقریباً نرمال و متغیر دسته‌ای دارای دو گروه مستقل باشد.
    Welch T-test: مثل T-test است، ولی زمانی استفاده می‌شود که واریانس دو گروه برابر نباشد.
    Mann–Whitney U: وقتی متغیر عددی غیرنرمال و متغیر دسته‌ای دارای دو گروه مستقل باشد.
    ANOVA: وقتی متغیر عددی تقریباً نرمال و متغیر دسته‌ای دارای سه گروه مستقل یا بیشتر باشد.
    Welch ANOVA: وقتی متغیر عددی تقریباً نرمال، گروه‌ها سه‌تا یا بیشتر و واریانس آن‌ها نابرابر باشد.
    Kruskal–Wallis: وقتی متغیر عددی غیرنرمال و متغیر دسته‌ای دارای سه گروه مستقل یا بیشتر باشد.

کتگوری به کتگوری

    Chi-square: برای بررسی ارتباط و استقلال دو متغیر دسته‌ای، زمانی که تعدادهای مورد انتظار کافی باشند.
    Fisher’s Exact: برای دو متغیر دسته‌ای با جدول 2×2، وقتی تعداد داده یا مقدارهای مورد انتظار کم باشد.
    McNemar: برای دو متغیر باینری وابسته، مانند پاسخ بله/خیر افراد یکسان قبل و بعد از یک اتفاق.

عددی به عددی

    Pearson: وقتی هر دو متغیر عددی، رابطه آن‌ها تقریباً خطی و داده پرت شدید نداشته باشند.
    Spearman: وقتی متغیرها غیرنرمال یا رتبه‌ای باشند و رابطه آن‌ها یکنواخت ولی لزوماً خطی نباشد.
    Kendall: برای متغیرهای رتبه‌ای، نمونه‌های کوچک یا داده‌هایی با رتبه‌های مساوی زیاد.

داده‌های وابسته یا قبل و بعد

    Paired T-test: برای مقایسه دو اندازه‌گیری وابسته، زمانی که اختلاف‌های آن‌ها تقریباً نرمال باشد.
    Wilcoxon Signed-Rank: برای مقایسه دو اندازه‌گیری وابسته و غیرنرمال.
    Repeated-Measures ANOVA: برای سه اندازه‌گیری وابسته یا بیشتر با شرایط نرمال.
    Friedman: برای سه اندازه‌گیری وابسته یا بیشتر با توزیع غیرنرمال.

اندازه اثر

    T-test: مقدار Cohen’s d
    Mann–Whitney U: مقدار Rank-biserial
    ANOVA: مقدار Eta Squared یا Omega Squared
    Kruskal–Wallis: مقدار Epsilon Squared
    Chi-square: مقدار Cramér’s V
    Pearson / Spearman / Kendall: خود ضریب همبستگی

- Cramér's V :
    - 0.10 → ارتباط ضعیف
    - 0.30 → ارتباط متوسط
    - 0.50 → ارتباط قوی

```python
from scipy.stats import chi2_contingency


def chi2_cramers(data , target , categorical_columns):

    categorical_columns = [
        column
        for column in categorical_columns
        if column != target
    ]

    number_of_tests = len(
        categorical_columns
    )

    results = []

    for column in categorical_columns:

        observed = pd.crosstab(
            data[target],
            data[column]
        )

        chi2, p_value, _, _ = (
            chi2_contingency(observed)
        )

        # اصلاح Bonferroni
        adjusted_p = min(
            p_value * number_of_tests,
            1
        )

        # محاسبه Cramér's V
        n = observed.to_numpy().sum()
        rows, columns = observed.shape

        cramers_v = np.sqrt(
            chi2 / (
                n * min(
                    rows - 1,
                    columns - 1
                )
            )
        )

        results.append(
            {
                "Variable": column,
                "P_Value": adjusted_p,
                "Cramers_V": cramers_v
            }
        )

    return pd.DataFrame(
        results
    ).sort_values(
        "Cramers_V",
        ascending=False
    )

categorical_columns = [
    "job",
    "marital",
    "education",
    "default",
    "housing",
    "loan",
    "contact",
    "month",
    "day_of_week",
    "poutcome"
]

results = chi2_cramers(
    data=data,
    target="deposit",
    categorical_columns=categorical_columns
)

results
```

- Kruskal–Wallis : 
    - فرض صفر: توزیع متغیر عددی در تمام گروه‌ها یکسان است.
    - Epsilon Squared : 
        - 0.01 → اثر ضعیف
        - 0.06 → اثر متوسط
        - 0.14 → اثر بزرگ
- U_test : 
    - 0.10	اثر ضعیف
    - 0.30	اثر متوسط
    - 0.50	اثر بزرگ

```python
from scipy.stats import kruskal, mannwhitneyu


def group_association(data, target, columns):
    """
    برای هر ستون، تست مناسب (U یا Kruskal)، P_Value با Bonferroni
    و اندازه‌ی اثر ارتباطش با target را برمی‌گرداند.
    """

    columns = [
        column
        for column in columns
        if column != target
    ]

    number_of_tests = len(columns)

    results = []

    for column in columns:

        clean = data[[target, column]].dropna()

        values = clean[column]
        labels = clean[target].to_numpy()

        groups = [values[labels == g] for g in pd.unique(labels)]
        groups = [g for g in groups if len(g) > 0]

        if len(groups) < 2:
            continue

        if len(groups) == 2:
            # -------- دو گروه: Mann-Whitney U --------
            U, p_value = mannwhitneyu(groups[0], groups[1])

            n1, n2 = len(groups[0]), len(groups[1])

            # همبستگی رانک‌بیسریال (معیار استاندارد ۲ گروهی)
            effect = abs(1 - (2 * U) / (n1 * n2))

            test_name, statistic = "Mann-Whitney U", U

        else:
            # -------- سه گروه به بالا: Kruskal-Wallis --------
            H, p_value = kruskal(*groups)

            n, k = len(clean), len(groups)
            effect = (H - k + 1) / (n - k)  # Epsilon-Squared

            test_name, statistic = "Kruskal-Wallis", H

        # اصلاح Bonferroni
        adjusted_p = min(p_value * number_of_tests, 1)

        results.append(
            {
                "Variable": column,
                "Test": test_name,
                "Statistic": statistic,
                "Bonferroni" : adjusted_p ,
                "P_Value": p_value ,
                "Effect_Size": effect,
            }
        )

    return pd.DataFrame(results).sort_values(
        "Effect_Size",
        ascending=False,
    )



results = group_association(
    data=data,
    target="deposit",                      # deposit
    columns=[
        "age",
        "duration",
        "campaign",
        "pdays",
        "previous",
        "eco_employment_rate_change_pct",
        "eco_consumer_price_index",
        "eco_consumer_confidence_index",
        "eco_interest_rate_3m_pct",
        "eco_num_employed_thousand",
    ]
)

results
```

- Cramér's V :
    - 0.10 → ارتباط ضعیف
    - 0.30 → ارتباط متوسط
    - 0.50 → ارتباط قوی

```python
from scipy.stats import chi2_contingency
from matplotlib.ticker import PercentFormatter

sns.set_theme(
    style="white",
    rc={
        "figure.facecolor": "#F7F5FC",
        "axes.facecolor": "#FFFFFF"
    }
)

categorical_columns = ["job","marital","education","default","housing","loan" , "new_age"]
 
number_of_tests = len(
    categorical_columns
)

colors = [
    "#FF5C7A",
    "#6D5DFB",
    "#00BFA6"
]

results = []


for column in categorical_columns:

    # جدول تعدادهای واقعی
    observed = pd.crosstab(
        data[column],
        data["poutcome"],
    )

    # آزمون Chi-square
    chi2, p_value, dof, expected = (
        chi2_contingency(observed)
    )

    # اصلاح Bonferroni
    adjusted_p = min(
        p_value * number_of_tests,
        1
    )

    # محاسبه Cramér's V
    n = observed.to_numpy().sum()
    rows, columns = observed.shape

    cramers_v = np.sqrt(
        chi2 / (
            n * min(
                rows - 1,
                columns - 1
            )
        )
    )

    # بررسی شرایط Expected
    expected_ok = (
        expected.min() >= 1
        and (expected < 5).mean() <= 0.20
    )

    results.append(
        {
            "Variable": column,
            "Adjusted_P": adjusted_p,
            "Cramers_V": cramers_v,
            "Expected_OK": expected_ok
        }
    )

    # تبدیل تعدادها به درصد سطری
    distribution = (
        observed.div(
            observed.sum(axis=1),
            axis=0
        )
        .mul(100)
        .rename(
            columns={
                "failure": "Previous Failure",
                "nonexistent": "No Previous Campaign",
                "success": "Previous Success"
            }
        )
        [
            [
                "Previous Failure",
                "No Previous Campaign",
                "Previous Success"
            ]
        ]
        .sort_values(
            "Previous Success"
        )
    )

    figure_height = max(
        5,
        len(distribution) * 0.55
    )

    fig, ax = plt.subplots(
        figsize=(14, figure_height)
    )

    distribution.plot(
        kind="barh",
        stacked=True,
        width=0.72,
        color=colors,
        edgecolor="white",
        linewidth=1.2,
        ax=ax
    )

    # نوشتن درصد داخل میله‌ها
    for container in ax.containers:

        labels = [
            f"{value:.1f}%"
            if value >= 4
            else ""
            for value in container.datavalues
        ]

        ax.bar_label(
            container,
            labels=labels,
            label_type="center",
            color="white",
            fontsize=9,
            fontweight="bold"
        )

    # نمایش درصد Success کنار هر میله
    for position, value in enumerate(
        distribution["Previous Success"]
    ):

        ax.text(
            101.5,
            position,
            f"{value:.1f}% success",
            va="center",
            color="#007F70",
            fontsize=9,
            fontweight="bold"
        )

    p_text = (
        "p < 0.001"
        if adjusted_p < 0.001
        else f"p = {adjusted_p:.3f}"
    )

    ax.set_title(
        (
            f"Previous Campaign Outcome Distribution by {column}\n"
            f"{p_text}  |  Cramér's V = {cramers_v:.3f}"
        ),
        fontsize=15,
        fontweight="bold",
        color="#25213B",
        pad=22
    )

    ax.text(
        0,
        1.015,
        f"Each bar represents 100% of customers in that {column} category",
        transform=ax.transAxes,
        color="#777187",
        fontsize=10
    )

    ax.set_ylabel(
        column,
        fontsize=11,
        fontweight="bold"
    )

    ax.set_xlim(0, 114)

    ax.set_xticks(
        np.arange(0, 101, 20)
    )

    ax.xaxis.set_major_formatter(
        PercentFormatter(xmax=100)
    )

    ax.legend(
        title="Previous Campaign",
        ncol=3,
        frameon=False,
        loc="upper center",
        bbox_to_anchor=(0.5, -0.10)
    )

    for spine in ax.spines.values():
        spine.set_visible(False)

    plt.tight_layout()

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency
from matplotlib.ticker import PercentFormatter

sns.set_theme(
    style="white",
    rc={
        "figure.facecolor": "#F7F5FC",
        "axes.facecolor": "#FFFFFF"
    }
)

# new_age هدفِ استکه، نه ردیف
categorical_columns = ["job", "marital", "education", "default",
                       "housing", "loan"]

age_order = ["17-25", "26-35", "36-50", "51-60", "61+"]
month_order = ["jan", "feb", "mar", "apr", "may", "jun",
               "jul", "aug", "sep", "oct", "nov", "dec"]

age_colors = ["#00BFA6", "#54A0FF", "#6D5DFB", "#FF9F43", "#FF5C7A"]

number_of_tests = len(categorical_columns)

results = []


for column in categorical_columns:

    # جدول تعدادهای واقعی: ردیف‌ها = دسته‌های متغیر، ستون‌ها = گروه‌های سنی
    observed = pd.crosstab(
        data[column],
        data["new_age"],
    )

    # آزمون Chi-square
    chi2, p_value, dof, expected = (
        chi2_contingency(observed)
    )

    # اصلاح Bonferroni
    adjusted_p = min(
        p_value * number_of_tests,
        1
    )

    # محاسبه Cramér's V
    n = observed.to_numpy().sum()
    rows, columns = observed.shape

    cramers_v = np.sqrt(
        chi2 / (
            n * min(
                rows - 1,
                columns - 1
            )
        )
    )

    # بررسی شرایط Expected
    expected_ok = (
        expected.min() >= 1
        and (expected < 5).mean() <= 0.20
    )

    results.append(
        {
            "Variable": column,
            "Adjusted_P": adjusted_p,
            "Cramers_V": cramers_v,
            "Expected_OK": expected_ok
        }
    )

    # تبدیل تعدادها به درصد سطری (ستون‌ها به ترتیب سنی)
    distribution = (
        observed.div(
            observed.sum(axis=1),
            axis=0
        )
        .mul(100)
        .reindex(columns=age_order)
        [age_order]
    )

    # month به ترتیب زمانی، بقیه مرتب‌شده بر اساس سهم 61+
    if column == "month":
        distribution = distribution.reindex(
            [m for m in month_order if m in distribution.index]
        )
    else:
        distribution = distribution.sort_values("61+")

    figure_height = max(
        5,
        len(distribution) * 0.55
    )

    fig, ax = plt.subplots(
        figsize=(14, figure_height)
    )

    distribution.plot(
        kind="barh",
        stacked=True,
        width=0.72,
        color=age_colors,
        edgecolor="white",
        linewidth=1.2,
        ax=ax
    )

    # نوشتن درصد داخل میله‌ها
    for container in ax.containers:

        labels = [
            f"{value:.1f}%"
            if value >= 4
            else ""
            for value in container.datavalues
        ]

        ax.bar_label(
            container,
            labels=labels,
            label_type="center",
            color="white",
            fontsize=9,
            fontweight="bold"
        )

    # نمایش درصد 61+ کنار هر میله
    for position, value in enumerate(
        distribution["61+"]
    ):

        ax.text(
            101.5,
            position,
            f"{value:.1f}% 61+",
            va="center",
            color="#C2185B",
            fontsize=9,
            fontweight="bold"
        )

    p_text = (
        "p < 0.001"
        if adjusted_p < 0.001
        else f"p = {adjusted_p:.3f}"
    )

    ax.set_title(
        (
            f"Age Group Distribution by {column}\n"
            f"{p_text}  |  Cramér's V = {cramers_v:.3f}"
        ),
        fontsize=15,
        fontweight="bold",
        color="#25213B",
        pad=22
    )

    ax.text(
        0,
        1.015,
        f"Each bar represents 100% of customers in that {column} category",
        transform=ax.transAxes,
        color="#777187",
        fontsize=10
    )

    ax.set_ylabel(
        column,
        fontsize=11,
        fontweight="bold"
    )

    ax.set_xlim(0, 114)

    ax.set_xticks(
        np.arange(0, 101, 20)
    )

    ax.xaxis.set_major_formatter(
        PercentFormatter(xmax=100)
    )

    ax.legend(
        title="Age Group",
        ncol=5,
        frameon=False,
        loc="upper center",
        bbox_to_anchor=(0.5, -0.10)
    )

    for spine in ax.spines.values():
        spine.set_visible(False)

    plt.tight_layout()

# خلاصه‌ی همه‌ی ۷ آزمون
results = pd.DataFrame(results)
results.sort_values(by="Cramers_V" , inplace=True , ascending=False)
results
```

```python
from scipy.stats import chi2_contingency
from matplotlib.ticker import PercentFormatter

sns.set_theme(
    style="white",
    rc={
        "figure.facecolor": "#F7F5FC",
        "axes.facecolor": "#FFFFFF"
    }
)

categorical_columns = ["job", "marital", "education", "contact",
                        "default", "housing", "loan"]

month_order = ["jan", "feb", "mar", "apr", "may", "jun",
               "jul", "aug", "sep", "oct", "nov", "dec"]

palette = [
    "#6D5DFB", "#FF5C7A", "#00BFA6", "#54A0FF",
    "#FF9F43", "#FFD166", "#06D6A0", "#A78BFA",
    "#F4A261", "#EF476F", "#118AB2", "#9D4EDD"
]

number_of_tests = len(categorical_columns)

results = []


for column in categorical_columns:

    # جدول تعدادهای واقعی (برای آزمون)
    observed = pd.crosstab(
        data[column],
        data["month"],
    )

    # آزمون Chi-square
    chi2, p_value, dof, expected = (
        chi2_contingency(observed)
    )

    # اصلاح Bonferroni
    adjusted_p = min(
        p_value * number_of_tests,
        1
    )

    # محاسبه Cramér's V
    n = observed.to_numpy().sum()
    rows, columns = observed.shape

    cramers_v = np.sqrt(
        chi2 / (
            n * min(
                rows - 1,
                columns - 1
            )
        )
    )

    # بررسی شرایط Expected
    expected_ok = (
        expected.min() >= 1
        and (expected < 5).mean() <= 0.20
    )

    results.append(
        {
            "Variable": column,
            "Adjusted_P": adjusted_p,
            "Cramers_V": cramers_v,
        }
    )

    # درصد سطری: ردیف‌ها = ماه‌ها (زمانی)، ستون‌ها = دسته‌های فاکتور
    # (دسته‌ها به ترتیب فراوانی کل → بزرگ‌ترین پایینِ استک)
    category_order = observed.sum(axis=1).sort_values(ascending=False).index

    distribution = (
        observed.T
        .div(observed.T.sum(axis=1), axis=0)
        .mul(100)
        [category_order]
        .reindex([m for m in month_order if m in observed.columns])
    )

    figure_height = max(
        5,
        len(distribution) * 0.55
    )

    fig, ax = plt.subplots(
        figsize=(14, figure_height)
    )

    distribution.plot(
        kind="barh",
        stacked=True,
        width=0.72,
        color=palette[:len(category_order)],
        edgecolor="white",
        linewidth=1.2,
        ax=ax
    )

    # نوشتن درصد داخل میله‌ها
    for container in ax.containers:

        labels = [
            f"{value:.1f}%"
            if value >= 4
            else ""
            for value in container.datavalues
        ]

        ax.bar_label(
            container,
            labels=labels,
            label_type="center",
            color="white",
            fontsize=9,
            fontweight="bold"
        )

    p_text = (
        "p < 0.001"
        if adjusted_p < 0.001
        else f"p = {adjusted_p:.3f}"
    )

    ax.set_title(
        (
            f"{column.replace('_', ' ').title()} Distribution by Month\n"
            f"{p_text}  |  Cramér's V = {cramers_v:.3f}"
        ),
        fontsize=15,
        fontweight="bold",
        color="#25213B",
        pad=22
    )

    ax.text(
        0,
        1.015,
        "Each bar represents 100% of customers contacted in that month",
        transform=ax.transAxes,
        color="#777187",
        fontsize=10
    )

    ax.set_ylabel(
        "month",
        fontsize=11,
        fontweight="bold"
    )

    ax.set_xlim(0, 105)

    ax.set_xticks(
        np.arange(0, 101, 20)
    )

    ax.xaxis.set_major_formatter(
        PercentFormatter(xmax=100)
    )

    # legend (تعداد ستون‌ها بر اساس تعداد دسته‌ها)
    k = len(category_order)
    ncol = 3 if k > 3 else k
    n_rows = int(np.ceil(k / ncol))

    handles, labels = ax.get_legend_handles_labels()
    fig.legend(
        handles, labels,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.005),
        ncol=ncol,
        frameon=False,
        fontsize=12,
        columnspacing=1.0,
        handletextpad=0.4,
        labelspacing=0.3,
        borderpad=0.3
    )

    for spine in ax.spines.values():
        spine.set_visible(False)

    plt.tight_layout(rect=(0, 0.04 + 0.055 * n_rows, 1, 1))

# توی کدوم شغل بیشترین نرخ سپرده گذاری رو داریم
display(data.groupby("job").deposit.mean().mul(100).reset_index().sort_values(by="deposit" , ascending=False))

# 31 درصد از کل دانشجو ها سپرده گذاری کردند
# 25 درصد از کل بازنشسته ها سپرده گذاری کردن

display(data.groupby("contact").deposit.mean().mul(100))
```

```python
from scipy.stats import chi2_contingency
from matplotlib.ticker import PercentFormatter

sns.set_theme(
    style="white",
    rc={
        "figure.facecolor": "#F7F5FC",
        "axes.facecolor": "#FFFFFF"
    }
)

categorical_columns = [
    "job",
    "marital",
    "education",
    "contact",
    "day_of_week",
    "default",
    "housing",
    "loan"
]

month_order = [
    "jan",
    "feb",
    "mar",
    "apr",
    "may",
    "jun",
    "jul",
    "aug",
    "sep",
    "oct",
    "nov",
    "dec"
]

palette = [
    "#6D5DFB",
    "#FF5C7A",
    "#00BFA6",
    "#54A0FF",
    "#FF9F43",
    "#FFD166",
    "#06D6A0",
    "#A78BFA",
    "#F4A261",
    "#EF476F",
    "#118AB2",
    "#9D4EDD"
]

number_of_tests = len(
    categorical_columns
)

overall_rate = (
    data["deposit"].mean() * 100
)

minimum_sample_size = 30

results = []
count_tables = {}


for column in categorical_columns:

    rates = data.pivot_table(
        index="month",
        columns=column,
        values="deposit",
        aggfunc="mean",
        observed=False
    ).mul(100)

    counts = pd.crosstab(
        data["month"],
        data[column]
    )

    category_order = (
        data[column]
        .value_counts()
        .index
    )

    available_months = [
        month
        for month in month_order
        if month in rates.index
    ]

    rates = rates.reindex(
        index=available_months,
        columns=category_order
    )

    counts = counts.reindex(
        index=available_months,
        columns=category_order,
        fill_value=0
    )

    count_tables[column] = counts

    # گروه‌های دارای کمتر از ۳۰ مشتری فقط از نمودار حذف می‌شوند
    plot_rates = rates.where(
        counts >= minimum_sample_size
    ).dropna(
        axis=1,
        how="all"
    )

    counts = counts[
        plot_rates.columns
    ]

    if plot_rates.empty:
        continue

    observed = pd.crosstab(
        data[column],
        data["deposit"]
    )

    chi2, p_value, dof, expected = (
        chi2_contingency(observed)
    )

    adjusted_p = min(
        p_value * number_of_tests,
        1
    )

    n = observed.to_numpy().sum()
    rows, columns_count = observed.shape

    cramers_v = np.sqrt(
        chi2 / (
            n * min(
                rows - 1,
                columns_count - 1
            )
        )
    )

    expected_ok = (
        expected.min() >= 1
        and (expected < 5).mean() <= 0.20
    )

    v_tag = (
        "negligible"
        if cramers_v < 0.10
        else "weak"
        if cramers_v < 0.30
        else "moderate"
        if cramers_v < 0.50
        else "strong"
    )

    results.append(
        {
            "Variable": column,
            "Adjusted_P": adjusted_p,
            "Cramers_V": cramers_v,
            "Expected_OK": expected_ok
        }
    )

    p_text = (
        "p < 0.001"
        if adjusted_p < 0.001
        else f"p = {adjusted_p:.3f}"
    )

    fig, ax = plt.subplots(
        figsize=(14, 7)
    )

    x = np.arange(
        len(plot_rates.index)
    )

    for category, color in zip(
        plot_rates.columns,
        palette
    ):

        y_values = plot_rates[
            category
        ].to_numpy(
            dtype=float
        )

        sample_sizes = counts[
            category
        ].to_numpy(
            dtype=float
        )

        valid = np.isfinite(
            y_values
        )

        ax.plot(
            x,
            y_values,
            linewidth=2,
            color=color,
            alpha=0.75,
            label=category,
            zorder=2
        )

        point_sizes = np.clip(
            np.log1p(
                sample_sizes[valid]
            ) * 18,
            25,
            170
        )

        ax.scatter(
            x[valid],
            y_values[valid],
            s=point_sizes,
            color=color,
            edgecolor="white",
            linewidth=1,
            zorder=3
        )

    ax.axhline(
        overall_rate,
        color="#25213B",
        linestyle="--",
        linewidth=1.5,
        alpha=0.65,
        zorder=1
    )

    ax.set_title(
        (
            f"{column.replace('_', ' ').title()}"
            f" — Deposit Rate by Month\n"
            f"{p_text} | "
            f"Cramér's V = {cramers_v:.3f} "
            f"({v_tag})"
        ),
        fontsize=15,
        fontweight="bold",
        color="#25213B",
        loc="left",
        pad=15
    )

    ax.text(
        0,
        1.015,
        (
            f"Marker size = sample size"
            f"   •   groups with n < {minimum_sample_size} are hidden"
            f"   •   dashed = overall rate ({overall_rate:.1f}%)"
        ),
        transform=ax.transAxes,
        color="#777187",
        fontsize=10
    )

    ax.set_xticks(x)

    ax.set_xticklabels(
        plot_rates.index,
        fontsize=10,
        color="#25213B"
    )

    ax.set_xlim(
        -0.55,
        len(plot_rates.index) - 0.45
    )

    ax.set_ylim(
        0,
        min(
            115,
            np.nanmax(
                plot_rates.values
            ) * 1.22
        )
    )

    ax.tick_params(
        axis="y",
        labelsize=10,
        length=0,
        colors="#777187"
    )

    ax.yaxis.set_major_formatter(
        PercentFormatter(xmax=100)
    )

    ax.grid(
        axis="y",
        color="#E5E1F0",
        linewidth=0.8
    )

    ax.grid(
        axis="x",
        visible=False
    )

    ax.set_axisbelow(True)

    for spine in ax.spines.values():
        spine.set_visible(False)

    category_count = len(
        plot_rates.columns
    )

    legend_columns = (
        3
        if category_count > 3
        else category_count
    )

    legend_rows = int(
        np.ceil(
            category_count
            / legend_columns
        )
    )

    handles, labels = (
        ax.get_legend_handles_labels()
    )

    fig.legend(
        handles,
        labels,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.005),
        ncol=legend_columns,
        frameon=False,
        fontsize=12,
        columnspacing=1,
        handletextpad=0.4
    )

    plt.tight_layout(
        rect=(
            0,
            0.04 + 0.045 * legend_rows,
            1,
            0.98
        )
    )

    plt.show()


results = pd.DataFrame(results)

results = results.sort_values(
    "Cramers_V",
    ascending=False
).reset_index(drop=True)

results
```

```python
from scipy.stats import mannwhitneyu
from matplotlib.ticker import PercentFormatter


month_order = [
    "mar",
    "apr",
    "may",
    "jun",
    "jul",
    "aug",
    "sep",
    "oct",
    "nov",
    "dec"
]

economic_variables = [
    (
        "eco_employment_rate_change_pct",
        "Employment Rate Change"
    ),
    (
        "eco_consumer_price_index",
        "Consumer Price Index"
    ),
    (
        "eco_consumer_confidence_index",
        "Consumer Confidence Index"
    ),
    (
        "eco_interest_rate_3m_pct",
        "3-Month Interest Rate"
    ),
    (
        "eco_num_employed_thousand",
        "Number of Employed People"
    )
]

level_colors = {
    "Low": "#54A0FF",
    "Medium": "#7B61FF",
    "High": "#FF5C7A"
}

minimum_sample_size = 30

number_of_tests = len(
    economic_variables
)

overall_rate = (
    data["deposit"].mean() * 100
)

results = []

sns.set_theme(
    style="whitegrid",
    rc={
        "figure.facecolor": "#F7F5FC",
        "axes.facecolor": "#FFFFFF"
    }
)

deposit_values = data["deposit"].dropna().unique()

if len(deposit_values) != 2:
    raise ValueError(
        "Column 'deposit' must contain exactly two groups."
    )

for column, title in economic_variables:

    plot_data = data[
        [
            "month",
            column,
            "deposit"
        ]
    ].dropna().copy()


    group_1 = plot_data.loc[
        plot_data["deposit"] == deposit_values[0],
        column
    ]

    group_2 = plot_data.loc[
        plot_data["deposit"] == deposit_values[1],
        column
    ]


    if len(group_1) == 0 or len(group_2) == 0:
        continue


    U, p_value = mannwhitneyu(
        group_1,
        group_2,
        alternative="two-sided",
        method="auto"
    )


    n1 = len(group_1)
    n2 = len(group_2)


    effect_size = abs(
        1 - (
            2 * U
        ) / (
            n1 * n2
        )
    )

    adjusted_p = min(
        p_value * number_of_tests,
        1
    )


    strength = (
        "negligible"
        if effect_size < 0.10
        else "weak"
        if effect_size < 0.30
        else "moderate"
        if effect_size < 0.50
        else "strong"
    )



    p_text = (
        "p < 0.001"
        if adjusted_p < 0.001
        else f"p = {adjusted_p:.3f}"
    )


    results.append(
        {
            "Variable": column,
            "Test": "Mann-Whitney U",
            "Statistic": U,
            "Adjusted_P": adjusted_p,
            "P_Value": p_value,
            "Effect_Size": effect_size,
            "Strength": strength
        }
    )


    economic_groups = pd.qcut(
        plot_data[column],
        q=3,
        duplicates="drop"
    )


    number_of_groups = len(
        economic_groups.cat.categories
    )


    if number_of_groups == 2:

        group_labels = [
            "Low",
            "High"
        ]

    else:

        group_labels = [
            "Low",
            "Medium",
            "High"
        ]


    plot_data["EconomicLevel"] = (
        economic_groups.cat.rename_categories(
            group_labels
        )
    )


    level_ranges = (
        plot_data
        .groupby(
            "EconomicLevel",
            observed=True
        )[column]
        .agg(
            min="min",
            max="max",
            count="count"
        )
        .reindex(group_labels)
    )

    deposit_rates = (
        plot_data
        .pivot_table(
            index="month",
            columns="EconomicLevel",
            values="deposit",
            aggfunc="mean",
            observed=True
        )
        .mul(100)
    )


    sample_sizes = pd.crosstab(
        plot_data["month"],
        plot_data["EconomicLevel"]
    )


    available_months = [
        month
        for month in month_order
        if month in deposit_rates.index
    ]


    deposit_rates = deposit_rates.reindex(
        index=available_months,
        columns=group_labels
    )


    sample_sizes = sample_sizes.reindex(
        index=available_months,
        columns=group_labels,
        fill_value=0
    )

    plot_rates = deposit_rates.where(
        sample_sizes >= minimum_sample_size
    )


    if plot_rates.notna().sum().sum() == 0:
        continue

    fig, ax = plt.subplots(
        figsize=(15, 8)
    )


    x = np.arange(
        len(plot_rates.index)
    )

    for level in group_labels:

        values = plot_rates[
            level
        ].to_numpy(
            dtype=float
        )


        counts = sample_sizes[
            level
        ].to_numpy(
            dtype=float
        )


        valid = np.isfinite(values)


        ax.plot(
            x[valid],
            values[valid],
            color=level_colors[level],
            linewidth=2.7,
            alpha=0.85,
            label=level,
            zorder=2
        )

        point_sizes = np.clip(
            np.log1p(
                counts[valid]
            ) * 18,
            35,
            170
        )


        ax.scatter(
            x[valid],
            values[valid],
            s=point_sizes,
            color=level_colors[level],
            edgecolor="white",
            linewidth=1.2,
            zorder=3
        )

        for position in np.where(valid)[0]:

            ax.annotate(
                f"{values[position]:.1f}%",
                xy=(
                    x[position],
                    values[position]
                ),
                xytext=(0, 8),
                textcoords="offset points",
                ha="center",
                fontsize=8,
                fontweight="bold",
                color=level_colors[level]
            )


    ax.axhline(
        overall_rate,
        color="#25213B",
        linestyle="--",
        linewidth=1.5,
        alpha=0.65,
        label=f"Overall ({overall_rate:.1f}%)"
    )

    range_text_parts = []

    for level in group_labels:

        min_value = level_ranges.loc[
            level,
            "min"
        ]

        max_value = level_ranges.loc[
            level,
            "max"
        ]

        count_value = level_ranges.loc[
            level,
            "count"
        ]


        if pd.isna(min_value):
            continue

        if (
            abs(min_value) >= 100
            or abs(max_value) >= 100
        ):

            range_text = (
                f"{min_value:,.1f}"
                f" – "
                f"{max_value:,.1f}"
            )

        elif (
            abs(min_value) >= 10
            or abs(max_value) >= 10
        ):

            range_text = (
                f"{min_value:.1f}"
                f" – "
                f"{max_value:.1f}"
            )

        else:

            range_text = (
                f"{min_value:.2f}"
                f" – "
                f"{max_value:.2f}"
            )


        range_text_parts.append(
            f"{level}: [{range_text}]  n={int(count_value):,}"
        )


    range_text = "   •   ".join(
        range_text_parts
    )

    ax.set_title(
        (
            f"Deposit Rate by Month and {title} Level\n"
            f"Mann-Whitney U: {p_text}"
            f"  |  Rank-Biserial = "
            f"{effect_size:.3f} "
            f"({strength})"
        ),
        fontsize=17,
        fontweight="bold",
        color="#25213B",
        pad=22
    )


    ax.text(
        0,
        1.015,
        range_text,
        transform=ax.transAxes,
        color="#55516D",
        fontsize=10,
        fontweight="bold"
    )

    ax.text(
        0,
        0.975,
        (
            "Mann-Whitney test uses the original continuous "
            "economic variable"
            f"   •   groups with n < {minimum_sample_size} "
            "are hidden only from the chart"
        ),
        transform=ax.transAxes,
        color="#88839A",
        fontsize=9
    )

    ax.set_xlabel(
        "Contact Month",
        fontsize=11,
        fontweight="bold"
    )

    ax.set_ylabel(
        "Deposit Rate (%)",
        fontsize=11,
        fontweight="bold"
    )


    ax.set_xticks(x)

    ax.set_xticklabels(
        plot_rates.index
    )


    ax.set_xlim(
        -0.55,
        len(plot_rates.index) - 0.45
    )

    maximum_rate = np.nanmax(
        plot_rates.to_numpy()
    )


    ax.set_ylim(
        0,
        min(
            100,
            max(
                20,
                maximum_rate * 1.22
            )
        )
    )


    ax.yaxis.set_major_formatter(
        PercentFormatter(
            xmax=100
        )
    )

    ax.grid(
        axis="y",
        linestyle="--",
        color="#DDD8EA",
        alpha=0.75
    )


    ax.grid(
        axis="x",
        visible=False
    )

    ax.legend(
        title=f"{title} Level",
        frameon=False,
        ncol=4,
        loc="upper center",
        bbox_to_anchor=(0.5, -0.11)
    )


    for spine in ax.spines.values():
        spine.set_visible(False)


    plt.tight_layout()



results = (
    pd.DataFrame(results)
    .sort_values(
        "Effect_Size",
        ascending=False
    )
    .reset_index(drop=True)
)

```

- چه عواملی باعث شدن افت یا افزایش شدیدی در ماه های مختلف داشته باشیم 

```python
from scipy.stats import chi2_contingency
from scipy.stats import mannwhitneyu

categorical_drivers = [
    "contact",
    "job",
    "default",
    "education"
]


economic_variables = [
    (
        "eco_employment_rate_change_pct",
        "Employment Rate Change"
    ),
    (
        "eco_consumer_price_index",
        "Consumer Price Index"
    ),
    (
        "eco_consumer_confidence_index",
        "Consumer Confidence Index"
    ),
    (
        "eco_interest_rate_3m_pct",
        "3-Month Interest Rate"
    ),
    (
        "eco_num_employed_thousand",
        "Number of Employed People"
    )
]


economic_columns = [
    column
    for column, title in economic_variables
]


driver_variables = (
    categorical_drivers
    + economic_columns
)

required_columns = [
    "month",
    "deposit"
] + driver_variables


missing_columns = [
    column
    for column in required_columns
    if column not in data.columns
]


if missing_columns:

    raise KeyError(
        f"These columns are missing from data: "
        f"{missing_columns}"
    )

month_order = [
    "mar",
    "apr",
    "may",
    "jun",
    "jul",
    "aug",
    "sep",
    "oct",
    "nov",
    "dec"
]


months_present = [
    month
    for month in month_order
    if month in data["month"].dropna().unique()
]

def categorical_difference(
    target_group,
    other_group
):

    temp = pd.DataFrame({

        "group": (
            ["Target"] * len(target_group)
            + ["Other"] * len(other_group)
        ),

        "value": pd.concat(
            [
                target_group,
                other_group
            ],
            ignore_index=True
        )

    })


    temp = temp.dropna()


    if temp.empty:
        return np.nan


    if temp["value"].nunique() < 2:
        return np.nan


    table = pd.crosstab(
        temp["group"],
        temp["value"]
    )


    if table.shape[0] < 2:
        return np.nan


    if table.shape[1] < 2:
        return np.nan


    chi2, _, _, _ = chi2_contingency(
        table
    )


    n = table.to_numpy().sum()


    denominator = (
        n
        * min(
            table.shape[0] - 1,
            table.shape[1] - 1
        )
    )


    if denominator == 0:
        return np.nan


    return np.sqrt(
        chi2 / denominator
    )


def numeric_difference(
    target_group,
    other_group
):

    target_group = pd.to_numeric(
        target_group,
        errors="coerce"
    ).dropna()


    other_group = pd.to_numeric(
        other_group,
        errors="coerce"
    ).dropna()


    if len(target_group) == 0:
        return np.nan


    if len(other_group) == 0:
        return np.nan


    try:

        u_statistic, _ = mannwhitneyu(
            target_group,
            other_group,
            alternative="two-sided"
        )


        effect = abs(
            2
            * u_statistic
            / (
                len(target_group)
                * len(other_group)
            )
            - 1
        )


        return effect


    except ValueError:

        return np.nan


display_names = {

    "contact":
        "Contact",

    "job":
        "Job",

    "default":
        "Default",

    "education":
        "Education",

    "eco_employment_rate_change_pct":
        "Employment Rate",

    "eco_consumer_price_index":
        "Consumer Price Index",

    "eco_consumer_confidence_index":
        "Consumer Confidence",

    "eco_interest_rate_3m_pct":
        "3-Month Interest Rate",

    "eco_num_employed_thousand":
        "Number Employed"
}


all_matrices = {}


for economic_column, economic_title in economic_variables:

    plot_data = data[
        required_columns
    ].copy()


    plot_data = plot_data.dropna(
        subset=[
            economic_column,
            "month",
            "deposit"
        ]
    ).copy()


    ranks = plot_data[
        economic_column
    ].rank(
        method="first"
    )


    plot_data["_economic_level"] = pd.qcut(
        ranks,
        q=3,
        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )

    scores = []

    for month in months_present:

        for variable in driver_variables:

            if variable == economic_column:

                scores.append({

                    "month":
                        month,

                    "variable":
                        variable,

                    "score":
                        np.nan

                })

                continue


            level_scores = []


            for level in [
                "Low",
                "Medium",
                "High"
            ]:


                level_data = plot_data[
                    plot_data["_economic_level"]
                    == level
                ]

                target = level_data[
                    level_data["month"] == month
                ][variable]


                other = level_data[
                    level_data["month"] != month
                ][variable]


                if variable in categorical_drivers:

                    score = categorical_difference(
                        target,
                        other
                    )


                else:

                    score = numeric_difference(
                        target,
                        other
                    )


                if pd.notna(score):

                    level_scores.append(
                        score
                    )


            if level_scores:

                final_score = np.mean(
                    level_scores
                )

            else:

                final_score = np.nan


            scores.append({

                "month":
                    month,

                "variable":
                    variable,

                "score":
                    final_score

            })


    matrix = (
        pd.DataFrame(scores)
        .pivot(
            index="month",
            columns="variable",
            values="score"
        )
    )

    variable_order = (

        matrix
        .mean(
            axis=0,
            skipna=True
        )
        .sort_values(
            ascending=False
        )
        .index

    )

    month_order_sorted = (

        matrix
        .mean(
            axis=1,
            skipna=True
        )
        .sort_values(
            ascending=False
        )
        .index

    )

    matrix = matrix.loc[
        month_order_sorted,
        variable_order
    ]


    all_matrices[
        economic_column
    ] = matrix

    display_matrix = matrix.rename(
        columns=display_names
    )

    plt.figure(
        figsize=(14, 7)
    )


    sns.heatmap(

        display_matrix,

        annot=True,

        fmt=".2f",

        cmap="YlOrRd",

        vmin=0,

        vmax=1,

        linewidths=0.5,

        cbar_kws={
            "label":
                "Difference Strength"
        }

    )


    plt.title(

        f"Which Variables Differ Most by Month?\n"
        f"Within {economic_title} Levels",

        fontsize=16,

        fontweight="bold",

        pad=15

    )


    plt.xlabel(
        "Variables — Strongest → Weakest",
        fontsize=12
    )


    plt.ylabel(
        "Months — Strongest → Weakest",
        fontsize=12
    )


    plt.xticks(
        rotation=35,
        ha="right"
    )


    plt.yticks(
        rotation=0
    )


    plt.tight_layout()
```
